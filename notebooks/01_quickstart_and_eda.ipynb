## 1. Thử nghiệm bộ sinh lỗi chính tả nhân tạo (Noise Synthesizer)

```python
synthesizer = VietnameseNoiseSynthesizer(seed=42)

sample_sentences = [
    "Hôm nay trời rất trong xanh và mát mẻ.",
    "Trường đại học Bách Khoa có bề dày lịch sử và truyền thống.",
    "Trí tuệ nhân tạo và học sâu đang phát triển vô cùng mạnh mẽ.",
    "Tôi rất thích đọc sách nghiên cứu vào mỗi buổi tối cuối tuần.",
    "Mô hình ViT5 giải quyết rất tốt bài toán sinh văn bản và sửa lỗi."
]

print("=== KẾT QUẢ TẠO NHIỄU (CLEAN -> NOISY) ===\n")
for s in sample_sentences:
    noisy = synthesizer.add_noise(s, noise_prob=0.3)
    print(f"Gốc : {s}")
    print(f"Nhiễu: {noisy}\n")
```

## 2. Thử nghiệm độ đo đánh giá (Metrics)

```python
refs = ["Tôi đi học đại học", "Hôm nay trời đẹp", "Học sâu rất thú vị"]
preds = ["Tôi đi học đai học", "Hôm nay trời đẹp", "Học sâu rất thú zi"]

print(f"Exact Match: {compute_exact_match(preds, refs) * 100:.2f}%")
print(f"CER        : {compute_cer(preds, refs) * 100:.2f}%")
print(f"WER        : {compute_wer(preds, refs) * 100:.2f}%")
```
